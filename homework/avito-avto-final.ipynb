```python
def generate_dynamic_boxplot_summary(df: pd.DataFrame):
    """Генерирует динамический Markdown-отчет по результатам анализа диаграмм размаха."""

    # Вычисление статистик
    price_median = df['price'].median()
    price_q1 = df['price'].quantile(0.25)
    price_q3 = df['price'].quantile(0.75)

    year_median = int(df['year'].median())
    year_q1 = int(df['year'].quantile(0.25))
    year_q3 = int(df['year'].quantile(0.75))

    mileage_median = df['mileage'].median()
    mileage_q1 = df['mileage'].quantile(0.25)
    mileage_q3 = df['mileage'].quantile(0.75)

    engine_median = df['engine_volume'].median()
    engine_q1 = df['engine_volume'].quantile(0.25)
    engine_q3 = df['engine_volume'].quantile(0.75)

    # Формирование Markdown текста с использованием f-string
    report_md = f""" > **Анализ распределений и выбросов числовых признаков с помощью диаграмм размаха**
    На этом дашборде представлены диаграммы размаха (box plots) для четырех ключевых числовых признаков. Этот тип визуализации позволяет оценить медиану, межквартильный размах (IQR, \"ящик\", в котором находятся 50% всех данных) и наличие выбросов (точки за пределами \"усов\").
    1.  **Цена автомобиля (вверху слева):**
    *   **Медианная цена** составляет **{price_median:,.0f} руб.**, при этом половина всех автомобилей на рынке стоит в диапазоне **от {price_q1:,.0f} до {price_q3:,.0f} руб.**

*   Огромное количество выбросов в верхней части графика указывает на присутствие в данных очень дорогих автомобилей (люкс, спорткары, редкие модели), которые сильно отличаются от основной массы предложений.
    2.  **Год выпуска (вверху справа):**

    *   **Медианный год выпуска — {year_median} г**, а 50% автомобилей были произведены между **{year_q1} и {year_q3} годами**.

    *   Выбросы в нижней части графика — это раритетные, коллекционные или просто очень старые автомобили, которые не являются типичными для вторичного рынка.

    3.  **Пробег (внизу слева):**

    *   **Медианный пробег** — **{mileage_median:,.0f} км**, что характерно для подержанных автомобилей. Половина машин имеет пробег **от {mileage_q1:,.0f} до {mileage_q3:,.0f} км**.

    *   Множество выбросов с высоким значением пробега может соответствовать автомобилям, которые использовались в коммерческих целях (например, в такси или для доставки).

    4.  **Объем двигателя (внизу справа):**

    *   **Медианный объем двигателя — {engine_median:.1f} л**, что является стандартом для большинства автомобилей массового сегмента. 50% данных лежат в узком диапазоне **от {engine_q1:.1f} до {engine_q3:.1f} л**.

    *   Выбросы в верхней части — это автомобили с мощными, многолитровыми двигателями, характерными для внедорожников, спорткаров и автомобилей премиум-класса.


    **Общий вывод:** Все четыре признака имеют асимметричное распределение и большое количество выбросов. При построении моделей машинного обучения эти выбросы необходимо будет обработать, чтобы они не искажали результаты.
    """

    # Отображение сгенерированного отчета
    display(Markdown(report_md))
```

```python
import re

def convert_engine_volume_to_numeric(engine_volume_str):
    """
    Преобразует строковое значение объема двигателя в числовое
    Примеры: ': 1.6 л' -> 1.6, ': 2.0 л' -> 2.0
    """
    if pd.isna(engine_volume_str):
        return None

    # Удаляем все кроме цифр и точки
    cleaned = re.sub(r'[^0-9.]', '', str(engine_volume_str))

    # Если после очистки строка пустая
    if not cleaned:
        return None

    try:
        # Преобразуем в float
        return float(cleaned)
    except ValueError:
        return None



def remove_colon(name):
    """
    Преобразует строковое значение в строкове без :
    Примеры: ': 4+' -> 4+, ': 2.0 л' -> 2.0
    """
    if name:
        cleaned = re.sub(r':', '', str(name))
        cleaned = cleaned.strip()
        return cleaned
```

## 4.1. Feature Engineering и создание пайплайнов

### Извлечение признака `horsepower`

Признак `modification` содержит текстовую информацию о двигателе, включая мощность в лошадиных силах (например, "1.6 MT (106 л.с.)"). Это крайне важная информация для предсказания цены. Мы извлечем её в отдельный числовой признак `horsepower`.

```python
horsepower_pattern = r'(\d+)\s*л\.с\.'

# Извлекаем значения в новый столбец 'horsepower'
X_train['horsepower'] = X_train['modification'].str.extract(horsepower_pattern)
X_test['horsepower'] = X_test['modification'].str.extract(horsepower_pattern)
train_df['horsepower'] = train_df['modification'].str.extract(horsepower_pattern)

# Преобразуем столбец в числовой тип (float, чтобы можно было использовать NaN)
# errors='coerce' превратит все, что не удалось преобразовать, в NaN (пропуски)
X_train['horsepower'] = pd.to_numeric(X_train['horsepower'], errors='coerce')
X_test['horsepower'] = pd.to_numeric(X_test['horsepower'], errors='coerce')
train_df['horsepower'] = pd.to_numeric(train_df['horsepower'], errors='coerce')

print(f"Доля незаполненных значений horsepower в train: {X_train['horsepower'].isna().sum() / X_train.shape[0] * 100:.2f} %")
print(f"Доля незаполненных значений horsepower в test: {X_test['horsepower'].isna().sum() / X_test.shape[0] * 100:.2f} %")
```

### Создание пайплайнов

Для оценки влияния различных подходов к обработке данных мы создадим два пайплайна:

1.  **`base_pipeline`**: Базовый пайплайн, который удаляет только самые "грязные" и избыточные признаки (`modification`, `description`, детальную геолокацию). Он использует простые методы для заполнения пропусков (медиана для чисел, мода для категорий) и `OneHotEncoder` для всех категориальных признаков. Этот пайплайн страдает от высокой размерности из-за кодирования высококардинальных признаков (таких как `model` и `region`).

2.  **`smart_pipeline`**: Улучшенный пайплайн. В дополнение к удалению "грязных" признаков, он применяет более продвинутые техники:
    *   **Раздельная обработка признаков**: Числовые, категориальные с высокой и низкой кардинальностью обрабатываются по-разному.
    *   **`TopCategoriesReplacer`**: Группирует редкие категории в одну, чтобы уменьшить шум и количество итоговых признаков.


```python
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from feature_engine.selection import DropFeatures
from feature_engine.encoding import RareLabelEncoder
from category_encoders import CountEncoder

from sklearn.base import BaseEstimator, TransformerMixin
import pandas as pd
import numpy as np

class TopCategoriesReplacer(BaseEstimator, TransformerMixin):
    """
    Оставляет только N самых частых категорий, остальные заменяет на 'replace_with'.
    Совместим со sklearn Pipeline.
    """
    def __init__(self, top_n=5, replace_with='Другое'):
        self.top_n = top_n
        self.replace_with = replace_with
        self.top_categories_ = {}

    def fit(self, X, y=None):
        X = pd.DataFrame(X).copy()
        for col in X.columns:
            top_cats = X[col].value_counts().nlargest(self.top_n).index
            self.top_categories_[col] = set(top_cats)
        return self

    def transform(self, X):
        X = pd.DataFrame(X).copy()
        for col in X.columns:
            if col in self.top_categories_:
                X[col] = np.where(X[col].isin(self.top_categories_[col]),
                                  X[col],
                                  self.replace_with)
        return X


class HorsepowerExtractor(BaseEstimator, TransformerMixin):
    """
    Извлекает лошадиные силы из текстового столбца 'modification'
    и добавляет новый числовой столбец 'horsepower'.
    """
    def __init__(self, source_column='modification', target_column='horsepower', pattern=r'(\d+)\s*л\.с\.'):
        self.source_column = source_column
        self.target_column = target_column
        self.pattern = pattern

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        # Применяем регулярное выражение
        X[self.target_column] = X[self.source_column].str.extract(self.pattern)
        # Преобразуем в float (или int)
        X[self.target_column] = pd.to_numeric(X[self.target_column], errors='coerce')
        return X


# Пайплайн 1: Базовый подход
base_pipeline = Pipeline([
    ('drop_features', DropFeatures(features_to_drop=[
        'modification', 'street', 'house', 'district', 'city', 'package', 'description'
    ])),
    ('impute_and_encode', ColumnTransformer([
        ('numeric', make_pipeline(
            SimpleImputer(strategy='median'),
            StandardScaler()),
            make_column_selector(dtype_include=np.number)
        ),
        ('categorical', make_pipeline(
            SimpleImputer(strategy='most_frequent'),
            OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
            make_column_selector(dtype_include='category')
        )
    ]))
])

# Пайплайн 2: Умный подход
smart_pipeline = Pipeline([
    ('horsepower_extraction', HorsepowerExtractor()) ,
    ('drop_features', DropFeatures(features_to_drop=[
        'modification', 'street', 'house', 'district', 'city', 'package', 'description'
    ])),
    ('impute_and_encode', ColumnTransformer([
        ('numeric', make_pipeline(
            SimpleImputer(strategy='median', add_indicator=True),
            StandardScaler()),
            ['year', 'mileage', 'engine_volume', 'horsepower']
        ),
        ('high_cardinality', make_pipeline(
            SimpleImputer(strategy='most_frequent'),
            TopCategoriesReplacer(top_n=5, replace_with='Другое'),
            OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
            ['region', 'brand', 'model']
        ),
        ('low_cardinality', make_pipeline(
            SimpleImputer(strategy='most_frequent'),
            OneHotEncoder(handle_unknown='ignore', sparse_output=False)
        ), ['transmission', 'engine_type', 'drive', 'body_type', 'steering_wheel', 'color', 'owners', 'condition', 'generation'])
    ]))
])
```

## 4.2. Фильтрация выбросов

Как показал разведочный анализ, числовые признаки (`price`, `mileage`, `year`) содержат значительное количество выбросов.

Мы применим простую, но стратегию: **удалим из обучающей выборки** данные, которые выходят за пределы 0.5 и 99.5 перцентилей по цене, пробегу и году выпуска меньшего чем 0.5 перцентиль.

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, r2_score

knn_pipeline = Pipeline([
    ('preprocessing', base_pipeline), # Используем smart_pipeline
    ('model', KNeighborsRegressor())
])

params = {
    'model__n_neighbors': [1, 2, 3, 4, 5, 7, 10, 15]
}

grid_knn = GridSearchCV(knn_pipeline, params, scoring='neg_root_mean_squared_error', cv=3, n_jobs=-1)
grid_knn.fit(X_train_clean, Y_train_clean)

best_k = grid_knn.best_params_['model__n_neighbors']
print(f"Лучшее значение k: {best_k}")
print(f"Лучший RMSE на кросс-валидации: {-grid_knn.best_score_:,.0f} руб.")

```

### Борьба с корреляцией (эксперимент, не вошедший в итоговую модель)

В ходе исследования мы также попробуем создать признаки, которые могли бы побороть мультиколлинеарность между исходными данными, например, `avg_mileage_per_year` (среднегодовой пробег) и `hp_per_liter` (удельная мощность).

#### Среднегодовой пробег
Признак `avg_mileage_per_year` был создан по формуле:
$$avg\_annual\_milleage = \frac{milleage}{2025 - year + 1}$$

Идея заключается в том, чтобы объединить два коррелирующих признака (`mileage` и `year`) в один. Однако, как показал анализ, корреляция этого нового признака с ценой оказалась практически нулевой.

**Причина:** У двух автомобилей с кардинально разной ценой (например, 3-летний и 10-летний) может быть одинаковый среднегодовой пробег. Таким образом, этот признак "смазывает" важную информацию о возрасте и общем износе, и его линейная связь с ценой теряется.

#### Удельная мощность
Признак `hp_per_liter` (лошадиных сил на литр) был создан для оценки технологичности двигателя:
$$hp\_per\_liter = \frac{horsepower}{engine\_volume}$$

**Наблюдение:** Корреляция `hp_per_liter` с ценой `(0.55)` оказалась ниже, чем у "сырого" признака `horsepower` `(0.68)`.

**Причина:** Этот признак является нелинейным. Высокое значение `hp_per_liter` может быть как у недорогого современного автомобиля с турбомотором малого объема, так и у очень дорогого спорткара. Простой коэффициент корреляции Пирсона не может уловить эту сложную связь.



```python
train_df['avg_annual_milleage'] = train_df['mileage'] / (2025 - train_df['year'] + 1)
train_df['hp_per_liter'] = train_df['horsepower'] / (train_df['engine_volume'] + 1e-10)
```

```python
# Анализ корреляции между числовыми признаками
plt.figure(figsize=(10, 8))
numeric_cols = ['avg_annual_milleage', 'hp_per_liter', 'price']
correlation_matrix = train_df[numeric_cols].corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0,
           square=True, linewidths=0.5, vmax=1, vmin=-1)
plt.title('Корреляция между новыми синтетическими признаками и ценой')
plt.tight_layout()
plt.show()
```
